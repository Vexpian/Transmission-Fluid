## Build and run your circuit

Build the circuit below and run it on Aer. Recover the key.

```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# Your circuit goes here.
circuit = None

qreg_q = QuantumRegister(25, 'q')
creg_c = ClassicalRegister(24, 'c')
circuit = QuantumCircuit(qreg_q, creg_c)

circuit.x(qreg_q[24])

for i in range(24):
    circuit.h(qreg_q[i])

oracle = helper.build_oracle()
circuit.append(oracle, range(25))

for j in range(24):
    circuit.h(qreg_q[j])

for k in range(24):
    circuit.measure(qreg_q[k], creg_c[k])

simulator = AerSimulator()
transpiled_circuit = transpile(circuit, simulator)
job = simulator.run(transpiled_circuit, shots=1)
result = job.result()
counts = result.get_counts()

measured_bitstring = list(counts.keys())[0]
s = int(measured_bitstring, 2)
print(f"Found integer s: {s}")



```
